```python
print("**O CHAPÉU SELETOR DE HOGWARTS**")

print("~Responda com honestidade para descobrir sua casa bruxa!~")


print("Você se considera mais corajoso, amistoso, inteligente, ambicioso ou nenhuma das opções?")

característica = input("Digite a sua resposta:").lower()

if característica == "corajoso":
    print("Você é da Grifinória!🦁")
elif característica == "amistoso":
    print("Você é da Lufa-Lufa!🦡")
elif característica == "inteligente":
    print("Você é da Corvinal!🦅")
elif característica == "ambicioso":
    print("Você é da Sonserina!🐍")
else:
    print("Você é trouxa!")
```
