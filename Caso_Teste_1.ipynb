```python
Orçamento = float (input( " Digite seu Orçamento ="  ) )
Urgência = int (input(" Digite sua Urgência de 1 a 5 ="  ))
if (Orçamento > 10000 and Urgência >=4):
    print (" Lead Vip")
elif(Orçamento > 5000 or Urgência == 5):
    print (" Lead Prioritário")
else:
   print (" Lead Padão")
```
